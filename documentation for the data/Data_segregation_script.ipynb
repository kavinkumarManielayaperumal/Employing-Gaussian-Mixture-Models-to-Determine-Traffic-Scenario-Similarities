```python
import pandas as pd

# Load the input file
tracks_df = pd.read_csv("/content/processed_traffic_data_with_matrix_and_time (5).csv")

# Split the data based on lane changes
lane_change_data = tracks_df[tracks_df['numLaneChanges'] > 0]  # Vehicles that changed lanes
no_lane_change_data = tracks_df[tracks_df['numLaneChanges'] == 0]  # Vehicles that didn't change lanes

# Take all lane change data (40%) and sample 60% from no lane change data
target_no_lane_change_count = int((len(lane_change_data) / 40) * 60)
no_lane_change_sampled = no_lane_change_data.sample(n=target_no_lane_change_count, random_state=42)

# Combine both datasets
final_combined_df = pd.concat([lane_change_data, no_lane_change_sampled], ignore_index=True)

# Save the final combined data to a new file
final_combined_df.to_csv("processed_tracks_with_laneChangeType.csv", index=False)

print("Data processing complete. The file has been saved as 'processed_tracks_with_laneChangeType.csv'.")
```
